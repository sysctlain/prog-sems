```python
from matplotlib import pyplot as plt

arr = np.random.uniform(0, 10, 10000)
x = arr[::2]
y = arr[1::2]
print(x[:10])
print("--------")
print(y[:10])
plt.scatter(x, y, s=5)

# plt.hist(x, np.arange(12) - 0.5) # чтобы значения бинов по оси были ровно посередине столбца

plt.scatter(x, y, s = 5)
#plt.vlines(np.arange(11) - 0.5, -1, 11, color="black", linestyle='--')
#plt.hlines(np.arange(11) - 0.5, -1, 11, color="black", linestyle='--')
plt.minorticks_on()
plt.xticks(np.arange(13)  - 0.5, "a b c d e f g")
plt.yticks(np.arange(13) - 0.5)
plt.xlim(0, 5)
plt.grid() # рисует только major, нештрихованно
plt.grid(which="minor", linestyle='--') # рисует только minor, причем штрихованно
# plt.show()
```

```python
from matplotlib import ticker

arr = np.random.normal(6, 1, size=3000)
x = arr[::3]
y = np.hstack([arr[1::3], arr[2::3]])

minor_ticker = ticker.AutoMinorLocator(4)
fig, axs = plt.subplots(1, 2, figsize=(7,4))
ax = axs[0]
# ax.scatter(x, y)
ax.minorticks_on()
ax.set_xticks(np.arange(13)  - 0.5, "a b c d e f g")
ax.set_yticks(np.arange(13) - 0.5)
ax.set_xlim(0, 5)
ax.hist(x, np.arange(13) - 0.5)
ax.xaxis.set_minor_locator(minor_ticker)
#ax.grid() # рисует только major, нештрихованно
#ax.grid(which="minor", linestyle='--') # рисует только minor, причем штрихованно

for ax in axs:
    ax.grid()
plt.subplots_adjust(wspace=0)
axs[1].yaxis.set_ticks(np.linspace(0, 60, 7), [])

plt.savefig("wow.svg")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(7, 4), sharey=700)

arr = np.random.normal(6, 1, size=3000)
size = arr.size
l_size = int(size * 4 / 9)
x = arr[:l_size]
y= arr[l_size:]
axs[0].hist(x, np.arange(13) - 0.5)
axs[1].hist(y, np.arange(13) - 0.5)
axs[0].set_title("woasjdsajdlksajdsasdjsalkjdksaljdsajdsakdjsakldslkajdslkadslkajdsakjdsalkjdalkjdasjkdalkdskjdjsadjksajdsakldlksjaw")
axs[1].set_title("?")
fig.suptitle("wow")
plt.subplots_adjust(wspace=0)

axs[0].yaxis.set_ticklabels(["$0$", r"$1 \cdot$"])
for ax in axs:
    ax.grid()
plt.savefig("wow.pdf")
```

```python
np.random.seed(42)
x = np.random.randn(100)

# compute a histogram by hand
bins = np.linspace(-5, 5, 20)
counts = np.zeros_like(bins)

# find the appropriate bin for each x
i = np.searchsorted(bins, x)
print(i)

# add 1 to each of these bins
np.add.at(counts, i, 1)
```

```python
plt.hist(x, bins, histtype='step');
```
