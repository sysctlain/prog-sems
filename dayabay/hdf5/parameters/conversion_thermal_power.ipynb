```python
n, r = map(int, input().split())
villages = list(map(int, input().split()))

for i in range(n):
    for j in range(n - 1 - i):
        if villages[j] > villages[j + 1]:
            villages[j], villages[j + 1] = villages[j + 1], villages[j]

count = 0
i = 0

while i < n:
    left = villages[i]

    while i < n and villages[i] <= left + r:
        i += 1

    tower = villages[i - 1]

    while i < n and villages[i] <= tower + r:
        i += 1

    count += 1

print(count)
```

```python
n = int(input())

all_results = []

for i in range(n):
    m = int(input())
    room = []

    for j in range(m):
        total, name = input().split()
        room.append((float(total), name))

    if not all_results:
        all_results = room
    else:
        merged = []
        i = 0
        j = 0

        while i < len(all_results) and j < len(room):
            score_all, name_all = all_results[i]
            score_room, name_room = room[j]

            if score_all > score_room or (
                score_all == score_room and name_all < name_room
            ):
                merged.append(all_results[i])
                i += 1
            else:
                merged.append(room[j])
                j += 1

        while i < len(all_results):
            merged.append(all_results[i])
            i += 1

        while j < len(room):
            merged.append(room[j])
            j += 1

        all_results = merged

print(len(all_results))

for total, name in all_results:
    print(f'{total:.2f} {name}')
```

.agg() — агрегировать

Получить одну строку на каждую группу.




.transform() — преобразовать

Получить результат для каждой исходной строки.




.apply() — применить свою функцию

Для более сложных операций над целой группой.

```python
import pandas as pd

df = pd.DataFrame({
    'city': ['A', 'A', 'B', 'B'],
    'name': ['Anna', 'Bob', 'Kate', 'Tom'],
    'sales': [100, 150, 80, 120],
    'cost': [60, 90, 50, 70],
    'category': ['food', 'tech', 'food', 'food']
})


result = df.groupby('city').agg(
    total_sales=('sales', 'sum'),
    total_cost=('cost', 'sum')
)

result['profit'] = result['total_sales'] - result['total_cost']

print(result)

#Здесь lambda x получает все значения столбца sales внутри одной группы.
# result = df.groupby('city').agg(
#     sales_range=('sales', lambda x: x.max() - x.min())
# )
```

```python
def calculate(group):
    return pd.Series({
        'profit': group['sales'].sum() - group['cost'].sum(),
        'ratio': group['sales'].sum() / group['cost'].sum()
    })

result = df.groupby('city').apply(calculate)

print(result)

# - В agg() функция обычно получает значения одного столбца группы.
# - В apply() функция получает всю таблицу одной группы (DataFrame).
# Поэтому можно обращаться к любым столбцам группы, сравнивать их, перемножать, вычислять сложные формулы.
```

```python
# Допустим, хотим по каждому городу:
# - посчитать средние продажи;
# - посчитать количество уникальных имён;
# - собрать все уникальные имена в одну строку;
# - определить наиболее частую категорию.
# Для этого можно использовать agg() и назначить каждому столбцу свою функцию.


def unique_names(x):
    return ', '.join(x.dropna().unique())

def most_common(x):
    modes = x.mode()

    if len(modes) == 0:
        return None

    return modes.iloc[0]


result = df.groupby('city').agg(
    avg_sales=('sales', 'mean'),
    name_count=('name', 'nunique'),
    all_names=('name', unique_names),
    popular_category=('category', most_common)
)

print(result)


# many funcs to one col
result = df.groupby('city').agg({
    'sales': ['mean', 'max', 'min'],
    'name': ['count', 'nunique'],
    'category': ['first', 'last', 'nunique']
})
```

Что хочешь сделать						Что использовать
Среднее, максимум, минимум по группам						agg()
Несколько функций по группам						agg()
Собственная функция для одного столбца						agg(lambda x: ...)
Функция зависит от нескольких столбцов группы						apply() или сначала agg(), потом расчёт
Добавить среднее по группе каждой исходной строке						transform()
Добавить сумму по группе каждой исходной строке						transform()
Выбрать определённые столбцы						[['col1', 'col2']]
Выбрать все числовые столбцы						select_dtypes(include='number')
Обработать текст особой функцией						agg() с собственной функцией

Группировать по нескольким признакам						groupby(['col1', 'col2'])

apply() лучше использовать, когда задачу нельзя удобно решить через agg() или transform(): он более универсален, но обычно менее эффективен.

```python
# разведка
n = int(input())
a = []

for i in range(n):
    l, r = map(int, input().split())
    a.append([l, r])

# Сортировка вставками по правой границе
for i in range(1, n):
    x = a[i]
    j = i - 1

    while j >= 0 and a[j][1] > x[1]:
        a[j + 1] = a[j]
        j -= 1

    a[j + 1] = x

count = 0
last = None

for l, r in a:
    if last is None or last < l:
        count += 1
        last = r

print(count)
```

```python
# анонимы
s = input()

odd_letters = []
even_letters = []

odd_pos = []
even_pos = []

counter = 0

for pos, ch in enumerate(s, 1):
    if ch.isupper():
        counter += 1

        if counter % 2 == 1:
            odd_letters.append(ch)
            odd_pos.append(str(pos))
        else:
            even_letters.append(ch)
            even_pos.append(str(pos))

print(''.join(odd_letters), ''.join(even_letters))
print(''.join(odd_pos), ''.join(even_pos))
```

```python
# комнаты
n = int(input())

rooms = []
total = 0

for i in range(n):
    m = int(input())
    room = []

    for j in range(m):
        score, name = input().split()

        # Храним баллы в целых сотых
        score = int(score.replace('.', ''))

        room.append((score, name))

    rooms.append(room)
    total += m

print(total)

pos = [0] * n

for _ in range(total):
    best = -1

    for i in range(n):
        if pos[i] < len(rooms[i]):

            if best == -1:
                best = i
            else:
                current = rooms[i][pos[i]]
                previous = rooms[best][pos[best]]

                if (current[0] > previous[0] or
                    (current[0] == previous[0] and
                     current[1] < previous[1])):
                    best = i

    score, name = rooms[best][pos[best]]
    pos[best] += 1

    sign = '-' if score < 0 else ''
    value = abs(score)

    formatted = f'{sign}{value // 100}.{value % 100:02d}'

    print(formatted, name)
```

```python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = list(grades)

    def add_grade(self, g):
        self.grades.append(g)

    @property
    def average(self):
        return sum(self.grades) / len(self.grades)


class ExcellentStudent(Student):
    @property
    def is_excellent(self):
        return self.average >= 90
```

```python
# М(ФТИ)
ru = 'абвгдеёжзийклмнопрстуфхцчшщъыьэюя'
en = 'abcdefghijklmnopqrstuvwxyz'


def shift(ch):
    for alphabet in (ru, en):
        lower = ch.lower()

        if lower in alphabet:
            i = alphabet.index(lower)
            new_ch = alphabet[(i - 1) % len(alphabet)]

            if ch.isupper():
                return new_ch.upper()

            return new_ch

    return ch


def encrypt(word):
    word = word[::-1]

    result = ''

    for ch in word:
        result += shift(ch)

    return result


words = input().split()

print(' '.join(encrypt(word) for word in words))
```

```python
from math import prod

def filter_tuples(arr: list[tuple[int, ...]], N: int) -> list[tuple[int, ...]]:
    return list(filter(lambda t: prod(t) == N, arr))
```
