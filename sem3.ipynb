```python
from math import *
for angle in range(0, 361, 15):
    print(f"{angle:?>3} | {sin(radians(angle)):+.5f}")
```

```python
def table():
    res = []
    for angle in range(0, 361, 15):
        # res.append(f"{angle:?>3} | {sin(radians(angle)):+.5f}")
        phi = radians(angle)
        res.append([angle, sin(phi), cos(phi)])
    return res
```

```python
tab = table()

with open("my_table.csv", "w") as f:
    for angle, s, c in tab:
        f.write(f"{angle:?>3} | {sin(radians(angle)):+.5f}\n")
```

```python
tab = table()

with open("my_table.csv", "w") as f:
    total = []
    for angle, s, c in tab:
        total.append(f"{angle:?>3},{sin(radians(angle)):+.5f},{cos(radians(angle)):+.5f}")
    f.write('\n'.join(total))

with open("my_table.csv", "r") as f:
    for row in f.readlines():
        angle, s, c = map(float, row.strip("\n").split(','))
        print(angle, s, c)
```
