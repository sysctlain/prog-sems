```python
# На вход подается массив из чисел. Вам необходимо отсортировать его так, чтобы нечетные по значению элементы остались на своих местах, а четные по значению элементы шли по возрастанию.
n = int(input())

a = [int(input()) for _ in range(n)]

even = []
for number in a:
    if number % 2 == 0:
        even.append(number)

for i in range(1, len(even)):
    current = even[i]
    j = i - 1

    while j >= 0 and even[j] > current:
        even[j + 1] = even[j]
        j -= 1

    even[j + 1] = current

index = 0
for i in range(n):
    if a[i] % 2 == 0:
        a[i] = even[index]
        index += 1

print(*a)
```

```python
# Дан список целых чисел. Отсортировать его так, чтобы сначала шли отрицательные по убыванию, потом — положительные во возрастанию.
numbers = list(map(int, input().split()))

negative = []
non_negative = []

for number in numbers:
    if number < 0:
        negative.append(number)
    else:
        non_negative.append(number)

for i in range(1, len(negative)):
    current = negative[i]
    j = i - 1

    while j >= 0 and negative[j] < current:
        negative[j + 1] = negative[j]
        j -= 1

    negative[j + 1] = current

for i in range(1, len(non_negative)):
    current = non_negative[i]
    j = i - 1

    while j >= 0 and non_negative[j] > current:
        non_negative[j + 1] = non_negative[j]
        j -= 1

    non_negative[j + 1] = current

print(*(negative + non_negative))



# ---

numbers = list(map(int, input().split()))

negative = [x for x in numbers if x < 0]
positive = [x for x in numbers if x >= 0]

print(
    ' '.join(map(str, sorted(negative, reverse=True))),
    ' '.join(map(str, sorted(positive)))
)

```

```python
#Дан список целых чисел. Отсортировать его по возрастанию, при этом оставить все нули на своих местах.

numbers = list(map(int, input().split()))

non_zero = []

for number in numbers:
    if number != 0:
        non_zero.append(number)

for i in range(1, len(non_zero)):
    current = non_zero[i]
    j = i - 1

    while j >= 0 and non_zero[j] > current:
        non_zero[j + 1] = non_zero[j]
        j -= 1

    non_zero[j + 1] = current

index = 0

for i in range(len(numbers)):
    if numbers[i] != 0:
        numbers[i] = non_zero[index]
        index += 1

print(*numbers)
```

```python
# Вы занимаетесь разработкой блог-сервиса, а именно реализуете разные базовые возможности. Одна из таких возможностей - поддержка списка друзей. Заданы: список пользователей, являющихся друзьями данного пользователя, и список пользователей, у которых данный пользователь содержится в списке друзей.
# Необходимо получить список друзей данного пользователя (Friends), список его взаимных друзей (Mutual Friends), и список тех пользователей, у кого данный пользователь содержится в списке друзей, но которые не являются его взаимными друзьями (Also Friend of).
import sys


def insertion_sort(items):
    """Сортировка вставками без использования sort/sorted."""
    result = items[:]

    for i in range(1, len(result)):
        current = result[i]
        j = i - 1

        while j >= 0 and result[j] > current:
            result[j + 1] = result[j]
            j -= 1

        result[j + 1] = current

    return result


data = sys.stdin.read().split()
position = 0

n = int(data[position])
position += 1

friends = []
for _ in range(n):
    friends.append(data[position])
    position += 1

m = int(data[position])
position += 1

also_friend_of = []
for _ in range(m):
    also_friend_of.append(data[position])
    position += 1

friends_set = set(friends)

mutual_friends = []
not_mutual_friends = []

for name in also_friend_of:
    if name in friends_set:
        mutual_friends.append(name)
    else:
        not_mutual_friends.append(name)

friends = insertion_sort(friends)
mutual_friends = insertion_sort(mutual_friends)
not_mutual_friends = insertion_sort(not_mutual_friends)


print("Friends: " + ", ".join(friends))
print("Mutual Friends: " + ", ".join(mutual_friends))
print("Also Friend of: " + ", ".join(not_mutual_friends))





# ---- 
n = int(input())
f = [input() for _ in range(n)]
m = int(input())
af = [input() for _ in range(m)]
mf = []
for i in list(f):
    if i in af:
        mf.append(i)
        af.remove(i)
print("Friends: " + ', '.join(sorted(f)))
print("Mutual Friends: " + ', '.join(sorted(mf)))
print("Also Friend of: " + ', '.join(sorted(af)))
```

```python
# Пусть задана строка s. Назовем ее k-ой (k > 0) степенью s^k строку s^k = sss (k раз). Например, третьей степенью строки abc является строка аbсаbсаbс.
s = input()
k = int(input())

if k > 0:
    print(s * k)
else:
    degree = -k

    if len(s) % degree != 0:
        print("NO SOLUTION")
    else:
        length = len(s) // degree
        root = s[:length]

        if root * degree == s:
            print(root)
        else:
            print("NO SOLUTION")
```

```python
# При сортировке слиянием одна из важных частей программы - функция, выполняющая слияние двух отсортированных массивов в больший отсортированный массив.
# 
# В данной задаче Вам необходимо написать функцию merge_lists(arr1, arr2), где arr1, arr2 -- два отсортированных по возрастанию списка.
# 
# В результате работы функция должна вернуть новый список, который содержит все элементы arr1, arr2. Все элементы должны быть отсортированы по возрастанию.
def merge_lists(arr1, arr2):
    result = []
    i = 0
    j = 0

    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1

    while i < len(arr1):
        result.append(arr1[i])
        i += 1

    while j < len(arr2):
        result.append(arr2[j])
        j += 1

    return result
```
