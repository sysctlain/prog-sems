```python
# Вам требуется реализовать класс PositiveCalendar. В позитивном календаре всего 13 месяцев, в каждом месяце 28 дней.

class PositiveCalendar:
    def __init__(self, ndays):
        self.year = (ndays) // (13 * 28)
        self.month = (ndays - self.year * 13 * 28)//28
        self.day = ndays - self.year * 13 * 28 - self.month * 28

    def print(self):
        print(f"Today is {self.year:0>4}-{self.month:0>2}-{self.day:0>2}")
```

```python
class Vector3D:
    def __init__(self, x, y, z):
        self._x = x
        self._y = y
        self._z = z

    def get_x(self):
        return self._x

    def get_y(self):
        return self._y

    def get_z(self):
        return self._z

    def abs(self):
        return (self._x ** 2 + self._y ** 2 + self._z ** 2) ** 0.5

    def vector_dot(self, other):
        return Vector3D(
            self._y * other._z - self._z * other._y,
            self._z * other._x - self._x * other._z,
            self._x * other._y - self._y * other._x
        )

    def scalar_dot(self, other):
        return self._x * other._x + self._y * other._y + self._z * other._z

    def transform(self, matrix):
        return Vector3D(
            matrix[0][0] * self._x + matrix[0][1] * self._y + matrix[0][2] * self._z,
            matrix[1][0] * self._x + matrix[1][1] * self._y + matrix[1][2] * self._z,
            matrix[2][0] * self._x + matrix[2][1] * self._y + matrix[2][2] * self._z
        )

    def print(self):
        print(f"{self._x:>+5.1f} {self._y:>+5.1f} {self._z:>+5.1f}")
```

```python
class PhoneBook:
    def __init__(self):
        self._book = {}

    def print(self, name: str) -> None:
        if name in self._book:
            print(f"{name}: {', '.join(self._book[name])}")

    def print_all(self) -> None:
        for name in sorted(self._book):
            self.print(name)

    def find(self, name: str) -> list[str]:
        return list(self._book.get(name, []))

    def add(self, name: str, phone: str) -> None:
        if name in self._book:
            self._book[name].append(phone)
        else:
            self._book[name] = [phone]

    def remove(self, name: str) -> None:
        if name in self._book:
            del self._book[name]

    def find_duplicates(self, number: str) -> tuple[str, ...]:
        return tuple(
            name for name, phones in self._book.items() if number in phones
        )
```

```python
class SimpleLinearClassifier:
    def __init__(self):
        self.k = 0.0
        self.b = 0.0
        self.memory = []

    def add(self, points):
        self.memory.extend(points)

    def fit(self):
        n = len(self.memory)
        if n == 0:
            self.k = 0.0
            self.b = 0.0
            return

        sum_x = sum(p[0] for p in self.memory)
        sum_y = sum(p[1] for p in self.memory)
        sum_xy = sum(p[0] * p[1] for p in self.memory)
        sum_x2 = sum(p[0] ** 2 for p in self.memory)

        denominator = n * sum_x2 - sum_x ** 2
        if denominator == 0:
            self.k = 0.0
            self.b = sum_y / n
        else:
            self.k = (n * sum_xy - sum_x * sum_y) / denominator
            self.b = (sum_y - self.k * sum_x) / n

    def predict(self, x):
        return self.k * x + self.b

    def reset(self):
        self.memory = []
```
