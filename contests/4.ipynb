```python
# Найти все вхождения строки A в строке B.
def main():
    A = input().strip()
    B = input().strip()

    n, m = len(A), len(B)

    if n > m:
        print(-1)
        return

    pi = [0] * n
    j = 0
    for i in range(1, n):
        while j > 0 and A[i] != A[j]:
            j = pi[j - 1]
        if A[i] == A[j]:
            j += 1
        pi[i] = j

    result = []
    j = 0
    for i in range(m):
        while j > 0 and B[i] != A[j]:
            j = pi[j - 1]
        if B[i] == A[j]:
            j += 1
        if j == n:
            result.append(i - n + 1)
            j = pi[j - 1]

    if result:
        print(" ".join(map(str, result)))
    else:
        print(-1)

main()


# ---

def compute_z_function(s):
    z = [0]
    left = right = 0
    for i in range(1, len(s)):
        x = min(z[i - left], right - i + 1) if i <= right else 0
        while i + x < len(s) and s[x] == s[i + x]:
            x += 1
        if i + x - 1 > right:
            left, right = i, i + x - 1
        z.append(x)
    return z


if __name__ == "__main__":
    pattern = input()
    text = input()
    z = compute_z_function(pattern + "#" + text)
    meets = []
    for i, v in enumerate(z):
        if v == len(pattern):
            meets.append(i-len(pattern)-1)
    if meets:
        print(" ".join(map(str, meets)))
    else:
        print("-1")
```

```python
# Нужно найти сколько раз A надо сдвинуть циклически влево, чтобы получить строку B. Циклический сдвиг влево на 1 превращает строку "abcde" в строку "bcdea".
A = input()
B = input()
n = len(A)

if len(B) != n:
    print(-1)
else:
    idx = (A + A).find(B)
    if idx != -1 and idx < n:
        print(idx)
    else:
        print(-1)
```

```python
# Напишите функцию convert_table_to_dict(data: str) -> dict[str, list[str] | list[int] | list[float]] преобразования передаваемого текста в словарь. Ключами словаря выступают названия столбцов. Значения за ключами являются списки. Если список содержит только числовые значения, то они должны быть преобразованы к типу int. Если же среди чисел встречаются точки, то данные должны быть преобразованы к float.

def convert_table_to_dict(data):
    lines = data.split()
    headers = lines[0].split(',')
    rows = [line.split(',') for line in lines[1:]]

    result = {}
    for i, header in enumerate(headers):
        column = [row[i] for row in rows]
        try:
            result[header] = [int(x) for x in column]
        except ValueError:
            try:
                result[header] = [float(x) for x in column]
            except ValueError:
                result[header] = column

    return result
```
