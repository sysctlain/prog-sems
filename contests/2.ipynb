```python
# На вход программе подается двумерный массив из N строк и M столбцов. Необходимо отсортировать значения в этом массиве так, каждая следующая строка представляла из себя отсортированный массив, минимальный элемент которого не меньше, час максимальный в пердыдущем. 


n = int(input())
m = int(input())

elements = []

for _ in range(n * m):
    elements.append(int(input()))

elements.sort()

for i in range(n):
    row = elements[i * m:(i + 1) * m]
    print(*row)
```

```python
# На вход программе подается массив чисел.Необходимо найти число, которое чаще всего встречается в массиве. Гарантируется, что такое число одно.
n = int(input())

counts = [0] * 100

for _ in range(n):
    number = int(input())
    counts[number] += 1

answer = counts.index(max(counts))

print(answer)
```

```python
# Вам необходимо написать программу, которая в данном потоке чисел заранее неизвестной длины находит максимальное значение, отбрасывая при этом последние 5 измерений.

from collections import deque

last_five = deque()
maximum = None

while True:
    number = int(input())

    if number == 0:
        break

    last_five.append(number)

    if len(last_five) > 5:
        value = last_five.popleft()

        if maximum is None or value > maximum:
            maximum = value

print(maximum)
```

```python
# Вам дано число N, которое не превышает 10000. Выведите все простые числа от 2 до N включительно. Операции умножения, деления и получение остатка использовать запрещено!
n = int(input())

is_prime = []
for n in range(n + 1):
    is_prime.append(True)
is_prime[0] = False
is_prime[1] = False

for number in range(2, n + 1):
    if is_prime[number]:
        print(number, end=" ")
        multiple = number + number

        while multiple <= n:
            is_prime[multiple] = False
            multiple += number
```

```python
# Дан двумерный массив. Необходимо вывести его элементы в порядке змейки (спиралью). 
n = int(input())
if n == 0:
    print(-1)

a = [[int(input()) for _ in range(n)] for _ in range(n)]

top, bottom = 0, n - 1
left, right = 0, n - 1

while top <= bottom and left <= right:
    for j in range(left, right + 1):
        print(a[top][j])
    top += 1

    for i in range(top, bottom + 1):
        print(a[i][right])
    right -= 1

    if top <= bottom:
        for j in range(right, left - 1, -1):
            print(a[bottom][j])
        bottom -= 1

    if left <= right:
        for i in range(bottom, top - 1, -1):
            print(a[i][left])
        left += 1
```

```python
#Вводится последовательность, состоящая только из 0 и 1. Необходимо найти максимальное количество 1, идущих подряд (без 0 между ними). 
def solve(N, arr):
    max_seq = 0
    cur_seq = 0
    for i in range(N):
        if arr[i] == 0:
            if cur_seq > max_seq:
                max_seq = cur_seq
            cur_seq = 0
        else:
            cur_seq += 1
    if cur_seq > max_seq:
        max_seq = cur_seq
    return max_seq
```

Есть результаты работы студентов в семестре. Студентов выводить в порядке суммы их баллов. Требуется вывести отсортированные результаты работ для каждого студента.

Данные вводятся как: student_id value

student_id принимает значения от 0 до N. value от 1 до 10

Пример входных данных: 0 3 0 5 1 3 1 2

Тут представленны данные о двух студента: 0 и 1. Сумма балов студента 0 - 8. Студента 1 - 5. Значит, сначала должны быть напечатаны результаты 0 студента, затем 1. Таким образом сначала надо вывести отсортированные результаты студента 0, затем студента 1:

5 3 3 2

Напомним, что у list в питоне есть встроенный метод sort и есть функция sorted. У них есть параметр key, который определяет по каким значениям будет сортироваться объект. Например код ниже будет сортировать лист по длинне его элементов. Так же есть параметр reverse.

            a = ['###', '@', '??'] a.sort(key=lambda x: len(x)) a ['@', '??', '###'] a.sort(key=lambda x: len(x), reverse=True) ['###', '??', '@']



```python
n = int(input())

students = [[] for _ in range(n)]

while True:
    s = input()
    if s == '#':
        break
    student_id, value = map(int, s.split())
    students[student_id].append(value)

students.sort(key=lambda x: sum(x), reverse=True)

result = []
for student in students:
    student.sort(reverse=True)
    result.extend(student)

print(*result)
```
