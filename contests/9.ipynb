```python
#Создайте датафрейм, в котором будует средние количества звонков Total day calls и Total eve calls для каждого штата.
df = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")

print(df.groupby(["State"]).agg({"Total day calls": "mean", "Total eve calls":"mean"}))
```

```python
#Создайте датафрейм, в котором будут средние количества звонков Total day calls и Total eve calls для каждого штата, выведите значение следующего выражения: mean(Total day calls) > mean(Total eve calls), где mean соответствует усреднению по группе штатов определенного столбца. 

f = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")
ff = f.groupby(by="State")[["Total day calls", "Total eve calls"]].mean()
print(ff["Total day calls"] > ff["Total eve calls"])

# -----

df = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")

df_means = df.groupby(["State"]).agg({"Total day calls": "mean", "Total eve calls":"mean"})

print(df_means["Total day calls"] > df_means["Total eve calls"])
```

```python
# Вывести DataFrame из 2 столбцов: число звонков в поддержку; доля оттока (Churn). Переименуйте столбец с долей оттока в Churn rate.

f = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")
print(f.groupby('Customer service calls').agg({'Churn': lambda x: x.mean()}).rename(columns={"Churn": "Churn rate"}))

# ----
df = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")
print(df.groupby(["Customer service calls"]).agg({"Churn":"mean"}).rename(columns={"Churn":"Churn rate"}))
```

```python
# Вывести DataFrame из 2 столбцов: число звонков в поддержку; число клиентов, звонивших столько раз. Столбец с ответом должен называться Number of clients.
f = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")

print(
    f.groupby('Customer service calls')
     .agg({'Churn': 'count'})
     .rename(columns={'Churn': 'Number of clients'})
)


# ----
df = pd.read_csv("http://python.mipt.ru/lpr/extra/lab08/telecom_churn.csv")

print(df.groupby(["Customer service calls"]).agg({"State":"count"}).rename(columns={"State":"Number of clients"}))
```
