```python
# Найдите доли выживших среди мужчин, среди женщин и среди всех пассажиров. Строка со средней долей выживших среди всех пасажиров должна иметь значение Gender равное all.
df = pd.read_csv("http://python.mipt.ru/lpr/extra/lab09/Titanic.csv")

ans = df.groupby("Gender").agg({"Survived": "mean"})

p = pd.DataFrame({"Survived": {"all": df["Survived"].mean()}})

p.index.name = "Gender"
answer = pd.concat([ans, p])


# ----

answer = df.groupby('Gender', as_index=False)['Survived'].mean() # <=> .agg(Surv: mean)

# строчка all
all_row = pd.DataFrame({
    'Gender': ['all'],
    'Survived': [df['Survived'].mean()]
})

answer = pd.concat([answer, all_row], ignore_index=True)

print(answer)
```
