Half life

Гордон Курчатов получил выплаты за очередное изобретение. Но в бухгалтерии все перепутали, ему выдали зарплату монетами неизвестного номинала, но одинакового размера. Количество монет равно N. В этот же день Гордон приступил к своему новому эксперименту.

Гордон измерил все монеты, убедился в их одинаковости, вычислил вероятность p выпадения орла и начал серию из M экспериментов. Он реализовал эксперименты по следующему алгоритму:

    Взять все монеты и подбросить каждую.
    Оставить монеты, которые выпали решкой вверх (0).
    Оставшиеся монеты, выпавшие орлами (1), убрать из эксперимента.
    Вернутся к пункту 1.

Напоминаем, что алгоритм повторяется M раз.

Постройте массив diff, в котором индекс соответствует итерации алгоритма, а значение по индексу - число выпавших орлов.

Пример: по нулевому индексу будет храниться количество изначальных монет N. По первому индексу N - [первое число решек]. По второму индексу N - [первое число решек] - [второе число решек]...

Выведите величину следующего формата: (diff[i] - diff[i - 1]) / diff[i]. Каждое значение в массиве должно быть выведено через пробел и с тремя знаками после точки.

Примечание: используйте np.random.binomial для получения вероятности выпадения орла.

```python
seed, N, p, M = input().split()
seed = int(seed)
N = int(N)
p = float(p)
M = int(M)

np.random.seed(seed)

diff = []

for _ in range(M):
    tosses = np.random.binomial(1, p, size=N)
    heads = np.sum(tosses)

    diff.append((N - heads) / N)
    N = heads

print(*[f"{x:.3f}" for x in diff])
```

В качестве ответа выведите три числа: значение максимального "икса", разность максимального "икса" с левой и правой границами полувысот "икса".
