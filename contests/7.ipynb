```python
# FT = I

N = int(input())
I = np.array(list(map(float, input().split())))
F = np.array([list(map(float, input().split())) for _ in range(N)])

T = np.linalg.solve(F, I)

print(", ".join(f"x{i} = {T[i]:.1f}" for i in range(N)))
```

```python
#Вам требуется вывести массив boolean значений, который соотвествует маске для выбора значений в массиве между lower_bound и upper_bound, но не достигает данных значений.
lower_bound, upper_bound = map(float, input().split())
data = np.array(list(map(float, input().split())))

result = np.logical_and(data > lower_bound, data < upper_bound)

print(" ".join(result.astype(str)))
```

```python
#Вам требуется заменить на среднее значение все места в массиве, которые меньше lower_bound и больше upper_bound. Среднее по массиву считается до применения маски!
lower_bound, upper_bound = map(float, input().split())
data = np.array(list(map(float, input().split())))

mean_value = data.mean()

mask = np.logical_or(data < lower_bound, data > upper_bound)
data[mask] = mean_value

print(*np.round(data, 1))
```

```python
#На вход программы подается двумерная матрица, которая считывается из файла matrix.dat. Определите степень ассимметричности матрицы.
A = np.loadtxt('matrix.dat', ndmin=2)

if A.shape[0] != A.shape[1]:
    print("It's imposibru")
else:
    numerator = (A - A.T) ** 2
    denominator = (A + A.T) ** 2

    fractions = np.divide(
        numerator,
        denominator,
        out=np.zeros_like(numerator, dtype=float),
        where=(denominator != 0)
    )

    gamma2 = 4 * np.sum(fractions)
    print(f'{gamma2:.2f}')

#
data = np.loadtxt("matrix.dat")

shape = data.shape

if shape[0] == shape[1]:
    mask = data + data.T != 0
    asymmetry = np.power(((data - data.T)[mask] / (data + data.T)[mask]), 2).sum()

    print(f"{4 * asymmetry:1.2f}")
else:
    print("It's imposibru")
```
