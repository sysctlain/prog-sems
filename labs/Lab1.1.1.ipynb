```python
from matplotlib import pyplot as plt
import numpy as np

plt.xlim([0, 180])
plt.ylim([0, 750])

def plot(x1, y1, x2, y2, i):
    colors = ["darkred", "purple", "green"]
    l = [50, 30, 20]

    X1 = x1
    Y1 = y1
    X2 = x2
    Y2 = y2
    X = np.concatenate((X1, X2))
    Y = np.concatenate((Y1, Y2))

    plt.errorbar(X2, Y2, marker="o", linestyle="none", xerr=0.05, yerr=2.5, color = 'blue', markersize=4)
    plt.errorbar(X1, Y1, marker="o", linestyle="none", xerr=0.05, yerr=2.5, color='red', markersize=4)

    k, b = np.polyfit(X, Y, 1)
    X = np.array([0, 180])
    print(k)
    plt.plot(X, k*X + b, color=colors[i], linestyle='-.', label=f"$l$ = {l[i]} см")



plot(I11, U11, I12, U12,0)
plot(I21, U21, I22, U22,1)
plot(I31, U31, I32, U32,2)

plt.title(r'$V(I)$') # добавляем заголовок к графику
plt.xlabel(r'$I$, мА') # добавляем подписи по оси x
plt.ylabel(r'$V$, мВ') # добавляем подписи по оси y
plt.minorticks_on()
plt.grid()
plt.grid(which="minor", linestyle=":")
plt.legend()
plt.savefig("plot.pdf")
plt.show()
```
